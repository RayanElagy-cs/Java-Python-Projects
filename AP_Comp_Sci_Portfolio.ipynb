```python
def quiz(questions, answers):
  points = 0
  responses = []

  for i in questions:
    ans = input(i).lower()
    responses.append(ans)

  for r, a in zip(responses, answers):
    if r == a:
      points+=1

  print("Your score is", (points/5)*100,"%")
quiz1questions = [
      "What is the 2nd year of high school called, (1 word) ",
      "What is soccer called in most countries ",
      "What country are frankfurts from ",
      "White hard tissue that makes up your skeletal system ",
      "Output device that displays information on a computer "
  ]

quiz1answers = ["sophomore", "football", "germany", "bones", "monitor"]

quiz2questions = [
    "What is 5+5 ",
    "What is 12^2 ",
    "What is 6/12 (fraction) ",
    "What is 3*5 ",
    "What is 10Log10(10000) (10Log base 10 (100)) "
 ]

quiz2answers = ["10", "144", "1/2", "15", "40"]
print("Quiz 1:")
quiz(quiz1questions, quiz1answers)
print("Quiz 2:")
quiz(quiz2questions, quiz2answers)
```
